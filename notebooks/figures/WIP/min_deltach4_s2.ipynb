```python
from marss2l.mars_sentinel2 import transmittance_to_ch4
t2ch4 = transmittance_to_ch4.TransmittanceCH4InterpolationFromDict()
t2ch4.amf_arr_max
```

```python
import matplotlib.pyplot as plt
import numpy as np

from importlib import reload
reload(transmittance_to_ch4)

amf_max = t2ch4.amf_arr_max
amf_min = 2
delta_toa = 1e-4

toa_b12 = np.arange(delta_toa, 0.7, delta_toa)
tr_b12 = 1 - delta_toa/toa_b12
delta_ch4_amf_min = t2ch4.deltach4_from_B12_transmittance("S2B", sza=None, vza=None, trans_b12=tr_b12, amf=amf_min)
delta_ch4_amf_max = t2ch4.deltach4_from_B12_transmittance("S2B", sza=None, vza=None, trans_b12=tr_b12, amf=amf_max)

delta_ch4_amf_min_b11 = t2ch4.deltach4_from_B11_transmittance("S2B", sza=None, vza=None, trans_b11=tr_b12, amf=amf_min)
delta_ch4_amf_max_b11 = t2ch4.deltach4_from_B11_transmittance("S2B", sza=None, vza=None, trans_b11=tr_b12, amf=amf_max)


fig, ax = plt.subplots(1,2,figsize=(8,4),tight_layout=True, sharey=True)
# ax[0].plot(toa_b12,delta_ch4_amf_min,label="min AMF")
# ax[0].plot(toa_b12,delta_ch4_amf_max,label="max AMF")
ax[0].fill_between(toa_b12, delta_ch4_amf_min, delta_ch4_amf_max, alpha=.7)

ax[1].fill_between(toa_b12, delta_ch4_amf_min_b11, delta_ch4_amf_max_b11, alpha=.7)
ax[0].set_ylim(0,700)
ax[0].set_title("B12")
ax[1].set_title("B11")
ax[0].set_ylabel(r"Minimum $\Delta$CH$_4$ that can be observed (ppb)")

for axi in ax:
    axi.set_xlabel("ToA reflectance")
    axi.set_xticks(np.arange(0,1.05,.1))
    axi.grid()
```

Radiometric resolution is routinely expressed as a bit number, typically in the range of 8 to 16 bits. The radiometric resolution of the MSI instrument is 12 bit[1], enabling the image to be acquired over a range of 0 to 4095 potential light intensity values. The radiometric accuracy is less than 5% (goal 3%). Radiometric resolution is also dependent upon the Signal to Noise Ratio (SNR) of the detector.


```python
fig, ax = plt.subplots(1,2,figsize=(8,4),tight_layout=True, sharey=True)
plt.yscale('log')

delta_ch4_amf_max_annual_report = t2ch4.deltach4_from_B12_transmittance("S2B", sza=None, vza=None, trans_b12=1-0.005, amf=amf_max)
delta_ch4_amf_min_annual_report = t2ch4.deltach4_from_B12_transmittance("S2B", sza=None, vza=None, trans_b12=1-0.005, amf=amf_min)
delta_ch4_amf_max_b11_annual_report = t2ch4.deltach4_from_B11_transmittance("S2B", sza=None, vza=None, trans_b11=1-0.005, amf=amf_max)
delta_ch4_amf_min_b11_annual_report = t2ch4.deltach4_from_B11_transmittance("S2B", sza=None, vza=None, trans_b11=1-0.005, amf=amf_min)
# ax[0].plot(toa_b12,delta_ch4_amf_min,label="min AMF")
# ax[0].plot(toa_b12,delta_ch4_amf_max,label="max AMF")
ax[0].fill_between(toa_b12, delta_ch4_amf_min, delta_ch4_amf_max, alpha=.7, label="Radimetric resolution")
ax[0].fill_between(toa_b12, 
                   np.repeat(delta_ch4_amf_min_annual_report, len(toa_b12)), 
                   np.repeat(delta_ch4_amf_max_annual_report, len(toa_b12)), alpha=.7, label="Quality assessment")

ax[1].fill_between(toa_b12, delta_ch4_amf_min_b11, delta_ch4_amf_max_b11, alpha=.7, label="Radimetric resolution")
ax[1].fill_between(toa_b12, 
                   np.repeat(delta_ch4_amf_min_b11_annual_report, len(toa_b12)), 
                   np.repeat(delta_ch4_amf_max_b11_annual_report, len(toa_b12)), alpha=.7, label="Quality assessment")
# ax[0].set_ylim(0,700)
ax[0].set_title("B12")
ax[1].set_title("B11")
ax[0].set_ylabel(r"Minimum $\Delta X$CH$_4$ (ppb)")

for axi in ax:
    axi.set_xlabel("ToA reflectance")
    axi.set_xticks(np.arange(0,1.05,.1))
    axi.grid()

axi.legend()

plt.savefig("figures/min_delta_ch4.pdf")
```
